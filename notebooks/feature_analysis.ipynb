### Objective:
Inspect the statistical separation between Alert and Drowsy classes across key temporal behavioral metrics: EAR, PERCLOS, blink duration, yawn count, and head nods.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, "..")
from config import FEATURES_CSV_PATH, FEATURE_COLUMNS

df = pd.read_csv(FEATURES_CSV_PATH)
df['label_name'] = df['label'].map({0: 'Alert', 1: 'Drowsy'})
df.head()
```

```python
# Key Features Boxplot Comparison (Alert vs Drowsy)
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
features_to_plot = ['EAR_mean', 'PERCLOS', 'avg_blink_duration', 'MAR_mean', 'yawn_count', 'head_nod_count']

for ax, feat in zip(axes.flatten(), features_to_plot):
    sns.boxplot(x='label_name', y=feat, data=df, ax=ax, palette=['#28a745', '#dc3545'])
    ax.set_title(f"{feat} by Driver State")
    ax.set_xlabel("State")
    ax.set_ylabel(feat)

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap of Temporal Behavioral Features
plt.figure(figsize=(12, 10))
corr = df[FEATURE_COLUMNS + ['label']].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, cbar=True)
plt.title("Correlation Matrix of Temporal Behavioral Features")
plt.tight_layout()
plt.show()
```
