### Objective:
Train and evaluate Random Forest, Support Vector Machine (SVM), and Logistic Regression on subject-aware partitions, generate confusion matrices, and evaluate feature importance.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, "..")
from config import FEATURES_CSV_PATH, FEATURE_COLUMNS, TARGET_COLUMN
from src.preprocessing import load_dataset, clean_dataset, subject_aware_split, scale_features
from src.evaluation import evaluate_model, plot_and_save_confusion_matrix, plot_feature_importance, save_comparison_table

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
```

```python
# Load and Clean Data
df = clean_dataset(load_dataset(FEATURES_CSV_PATH))
# Subject-Aware Partition
X_train, X_test, y_train, y_test, tr_subs, te_subs = subject_aware_split(df)
# Scale Features
X_tr_sc, X_te_sc, scaler = scale_features(X_train, X_test)
print(f"Training instances: {X_tr_sc.shape[0]}, Test instances: {X_te_sc.shape[0]}")
```

```python
# 1. Train and Evaluate Random Forest (Primary Model)
rf = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
rf.fit(X_tr_sc, y_train)
rf_res = evaluate_model(rf, X_te_sc, y_test.values, model_name="Random Forest")
```

```python
# 2. Train and Evaluate Support Vector Machine
svm = SVC(kernel='rbf', probability=True, random_state=42)
svm.fit(X_tr_sc, y_train)
svm_res = evaluate_model(svm, X_te_sc, y_test.values, model_name="SVM (RBF)")
```

```python
# 3. Train and Evaluate Logistic Regression Baseline
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_tr_sc, y_train)
lr_res = evaluate_model(lr, X_te_sc, y_test.values, model_name="Logistic Regression")
```

```python
# Comparison Table
comp_df = save_comparison_table([rf_res, svm_res, lr_res])
comp_df
```
