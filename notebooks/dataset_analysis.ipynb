### Objective:
Analyze the distribution of extracted facial and temporal features, check subject-level distribution, and verify the absence of missing values or identity leakage.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure project root is accessible
sys.path.insert(0, "..")
from config import FEATURES_CSV_PATH

print(f"Loading dataset from: {FEATURES_CSV_PATH}")
df = pd.read_csv(FEATURES_CSV_PATH)
df.head()
```

```python
# Dataset Shape and Data Types
print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns")
df.info()
```

```python
# Ground Truth Class Distribution
class_counts = df['label'].value_counts()
print("Ground Truth Labels:")
print(class_counts)

plt.figure(figsize=(6, 4))
sns.barplot(x=class_counts.index, y=class_counts.values, palette=["#28a745", "#dc3545"])
plt.title("Target Class Distribution (0 = Alert, 1 = Drowsy)")
plt.xlabel("Class Label")
plt.ylabel("Temporal Windows")
plt.show()
```

```python
# Subject-Level Representation
subject_counts = df['subject_id'].value_counts()
print(f"Total Unique Subjects: {df['subject_id'].nunique()}")

plt.figure(figsize=(10, 4))
sns.barplot(x=subject_counts.index, y=subject_counts.values, color="#4a90e2")
plt.title("Extracted Windows per Subject")
plt.xlabel("Subject ID")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Missing Values Assessment
missing_summary = df.isna().sum()
print("Missing Value Count per Column:")
print(missing_summary[missing_summary > 0] if missing_summary.sum() > 0 else "No missing values detected.")
```
